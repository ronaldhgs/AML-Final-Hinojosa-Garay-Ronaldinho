## A.1 Resumen Ejecutivo

El presente proyecto aborda la **optimización del pricing en Préstamos con Garantía Hipotecaria (PGH)** mediante técnicas de Machine Learning y Deep Learning.

**Situación actual:** La fijación de la Tasa Efectiva Anual (TEA) se realiza mediante un tarifario estático que agrupa clientes en segmentos predefinidos (por tipo de crédito, rango de score, etc.), asignando una misma tasa a perfiles que en realidad tienen niveles de riesgo muy diferentes. Este enfoque genera un error promedio de **2.11 puntos porcentuales** respecto a la tasa óptima individual.

**Problema central:** La lógica de pricing por "cajones" o segmentos no captura la granularidad del riesgo individual, lo que produce tres efectos negativos simultáneos:

1. **Pérdida de clientes valiosos:** Perfiles de bajo riesgo reciben tasas genéricas más altas de lo necesario, haciéndonos menos competitivos frente a alternativas que sí personalizan.
2. **Margen subóptimo:** Perfiles de alto riesgo pueden recibir tasas que no compensan adecuadamente su probabilidad de deterioro, erosionando la rentabilidad ajustada por riesgo del portafolio.
3. **Fricción operativa:** El proceso actual requiere iteraciones manuales, negociaciones ad-hoc y aprobaciones que consumen tiempo y recursos.

**Solución implementada:** Se desarrollaron tres modelos predictivos (XGBoost, Red Neuronal Profunda, Red Neuronal con Embeddings Categóricos) entrenados sobre un dataset de 3,800 contratos históricos con 160 variables, incluyendo comportamiento de pago mes a mes (cosechas). El mejor modelo (XGBoost) logró un **R² de 0.9485** y un error promedio de solo **0.37 pp de TEA**, representando una **mejora del 74.8%** sobre el tarifario estático.

**Resultado:** Un tarifario dinámico que genera una tasa personalizada por cada perfil individual, donde:
- Un cliente de bajo riesgo (score 780, LTV 35%) recibe **9.04%** en vez de un genérico ~12%.
- Un cliente de alto riesgo (score 480, LTV 75%) recibe **18.34%**, reflejando adecuadamente su perfil.

## A.3 Fuentes y Bibliografía

### A.3.1 Fundamentos de Machine Learning y Deep Learning

1. **Goodfellow, I., Bengio, Y., & Courville, A.** (2016). *Deep Learning*. MIT Press. — Referencia fundamental sobre arquitecturas de redes neuronales profundas, regularización (Dropout, L2), optimización (Adam) y batch normalization aplicados en los modelos DNN del proyecto. Disponible en: https://www.deeplearningbook.org/

2. **Chen, T., & Guestrin, C.** (2016). *XGBoost: A Scalable Tree Boosting System*. Proceedings of the 22nd ACM SIGKDD International Conference on Knowledge Discovery and Data Mining, 785-794. — Paper original de XGBoost, el modelo baseline utilizado, describiendo el algoritmo de gradient boosting con regularización. DOI: 10.1145/2939672.2939785

3. **Géron, A.** (2022). *Hands-On Machine Learning with Scikit-Learn, Keras, and TensorFlow* (3rd ed.). O'Reilly Media. — Guía práctica para implementación de pipelines de ML/DL, incluyendo feature engineering, validación cruzada y técnicas de regularización.

### A.3.2 Embeddings Categóricos y Datos Tabulares

4. **Guo, C., & Berkhahn, F.** (2016). *Entity Embeddings of Categorical Variables*. arXiv:1604.06737. — Fundamento teórico de los embeddings categóricos implementados en el Modelo 3, demostrando que las representaciones aprendidas capturan relaciones semánticas entre categorías. Disponible en: https://arxiv.org/abs/1604.06737

5. **Shwartz-Ziv, R., & Armon, A.** (2022). *Tabular Data: Deep Learning is Not All You Need*. Information Fusion, 81, 84-90. — Evidencia empírica de que XGBoost frecuentemente supera a Deep Learning en datos tabulares con volúmenes moderados, consistente con los resultados obtenidos en este proyecto. DOI: 10.1016/j.inffus.2021.11.011

### A.3.3 Interpretabilidad y SHAP

6. **Lundberg, S. M., & Lee, S. I.** (2017). *A Unified Approach to Interpreting Model Predictions*. 7. **Molnar, C.** (2022). *Interpretable Machine Learning: A Guide for Making Black Box Models Explainable* (2nd ed.). — Referencia complementaria sobre técnicas de interpretabilidad incluyendo SHAP, importancia de features y partial dependence plots. Disponible en: https://christophm.github.io/interpretable-ml-book/
